```python
team_mapping = {
    'Brighton & Hove Albion': 'Brighton',
    'Luton Town': 'Luton',
    'Leeds United': 'Leeds',
    'Leicester City': 'Leicester',
    'Manchester City': 'Man City',
    'Manchester United': 'Man United',
    'Newcastle United': 'Newcastle',
    'Norwich City': 'Norwich',
    'Nottingham Forest': 'Nottingham',
    'Sheffield United': 'Sheffield',
    'Tottenham Hotspur': 'Tottenham',
    'West Bromwich Albion': 'West Brom',
    'West Ham United': 'West Ham',
    'Wolverhampton Wanderers': 'Wolves'
}
```

```python
df2['team'] = df2['team'].replace(team_mapping)
```

```python
df2['date'] = pd.to_datetime(df2['date'])

df2['Year'] = df2['date'].dt.year
df2['Month'] = df2['date'].dt.month
df2['Day'] = df2['date'].dt.day
```

```python
df2['time'] = df2['time'].str.replace(r'\s*\(.*\)', '', regex=True).str.strip()
```

```python
df2.drop('comp', axis=1, inplace=True)
```

```python
df2.head()
```

```python
df2['round'] = df2['round'].str.extract(r'(\d+)').astype(int)
```

```python
def swap_teams(row):
    if row['venue'] == 'Home':
        row['opponent'], row['team'] = row['team'], row['opponent']
    return row

df2 = df2.apply(swap_teams, axis=1)
```

```python
def swap_teams_goals(row):
    if row['venue'] == 'Away':
        row['gf'], row['ga'] = row['ga'], row['gf']
    return row

df2 = df2.apply(swap_teams_goals, axis=1)
```

```python
df2 = df2.rename(columns={'opponent': 'Team1', 'team': 'Team2'})
```

```python
df2 = df2.rename(columns={'gf': 'GoalHome', 'ga': 'GoalAway'})
```

```python
df2 = df2.rename(columns={'xg': 'XGHome', 'xga': 'XGAway'})
```

```python
def check_possession(row):
    if row['venue'] == 'Away':
        row['poss'] = 100 - row['poss']
    return row
    
df2 = df2.apply(check_possession, axis=1)
```

```python
df2 = df2.rename(columns={'poss': 'PossessionHome'})
```

```python
df['match_key'] = df['HomeTeam'] + '_' + df['AwayTeam'] + '_' + df['Year'].astype(str) + '_' + df['Month'].astype(str) + '_' + df['Day'].astype(str)
df2['match_key'] = df2['Team1'] + '_' + df2['Team2'] + '_' + df2['Year'].astype(str) + '_' + df2['Month'].astype(str) + '_' + df2['Day'].astype(str)
```

```python
columns_to_add = ['referee', 'time', 'round', 'day', 'XGHome', 'XGAway', 'PossessionHome', 'attendance']
existing_columns = [col for col in columns_to_add if col in df2.columns]

df_merged = df.merge(df2[['match_key'] + existing_columns], 
                     on='match_key', 
                     how='left', 
                     suffixes=('', '_df2'))
```

```python
df['Referee'] = df['Referee'].fillna(df_merged['referee'])

for col in existing_columns:
    if col != 'referee':
        df[col] = df_merged[col + '_df2'] if col + '_df2' in df_merged.columns else df_merged[col]

if 'day' in df.columns:
    df = df.rename(columns={'day': 'WeekDay'})

df = df.drop([col for col in df.columns if col.endswith('_df2')], axis=1)
```
