```python
# Season statistics for Bolton Wanderers in 2003-04
bolton_stats = {
    'SeasonPosAway': 8,
    'SeasonPlayedAway': 38,
    'SeasonWinAway': 14,
    'SeasonDrawAway': 11,
    'SeasonLossAway': 13,
    'SeasonGFAway': 48,
    'SeasonGAAway': 56,
    'SeasonGDAway': -8,
    'SeasonPointsAway': 53,
    'QualificationOrRelegationAway': 'Not Applicable'
}

# Season statistics for Birmingham in 2003-04
Birmingham_stats = {
    'SeasonPosAway': 10,
    'SeasonPlayedAway': 38,
    'SeasonWinAway': 12,
    'SeasonDrawAway': 14,
    'SeasonLossAway': 12,
    'SeasonGFAway': 43,
    'SeasonGAAway': 48,
    'SeasonGDAway': -5,
    'SeasonPointsAway': 50,
    'QualificationOrRelegationAway': 'Not Applicable'
}

# List of columns to fill
away_cols = [
    'SeasonPosAway', 'SeasonPlayedAway', 'SeasonWinAway', 'SeasonDrawAway',
    'SeasonLossAway', 'SeasonGFAway', 'SeasonGAAway', 'SeasonGDAway',
    'SeasonPointsAway', 'QualificationOrRelegationAway'
]

# Fill the columns for all rows where AwayTeam is 'Bolton' in the 2003-04 season
for idx, row in df[df['Season'] == '0304'].iterrows():
    if row['AwayTeam'] == 'Bolton':
        for col in away_cols:
            df.loc[(df['Season'] == '0304') & (df.index == idx), col] = bolton_stats[col]

# Fill the columns for all rows where AwayTeam is 'Birmingham' in the 2003-04 season
for idx, row in df[df['Season'] == '0304'].iterrows():
    if row['AwayTeam'] == 'Birmingham':
        for col in away_cols:
            df.loc[(df['Season'] == '0304') & (df.index == idx), col] = Birmingham_stats[col]
```
