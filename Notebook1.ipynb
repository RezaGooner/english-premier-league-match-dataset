```python
import pandas as pd
import glob
import os
from pathlib import Path

pd.set_option('display.max_columns', None)
```

```python
pd.read_csv("RawData/season-2021.csv").head(10)
```

```python
folder_path = "RawData/" 
```

```python
all_files = glob.glob(os.path.join(folder_path, "*.csv"))
all_files.sort()
len(all_files)
```

```python
df_list = []

for file in all_files:
    df = pd.read_csv(file)
    file_name = os.path.basename(file).replace('.csv', '')
    df['Season'] = file_name  
    df_list.append(df)
    print(f"Read: {file_name}  |  Records: {len(df)}")

df = pd.concat(df_list, ignore_index=True)
```

```python
pd.read_csv("RawData/season-9394.csv")['HomeTeam'].nunique(), pd.read_csv("RawData/season-9495.csv")['HomeTeam'].nunique()
```

```python
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')
df = df.sort_values(by='Date').reset_index(drop=True)
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df.tail()
```

```python
new_order = ['Year', 'Month', 'Day'] + [col for col in df.columns if col not in ['Year', 'Month', 'Day', 'Date']]

df = df[new_order]

df.tail()
```
