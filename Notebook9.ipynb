```python
def get_stadium(team, year):
    if pd.isna(team) or pd.isna(year):
        return pd.NA

    try:
        year = int(year)
    except (ValueError, TypeError):
        return pd.NA

    matched_rows = stadium_history_df[
        (stadium_history_df['plus'] == team)
        & (stadium_history_df['StartYear'] <= year)
        & (stadium_history_df['EndYear'] >= year)
    ]

    if matched_rows.empty:
        return pd.NA

    return matched_rows.iloc[0]['Stadium']
```

```python
df = df.sort_values(['Year', 'Month', 'Day']).reset_index(drop=True)

team_history = {}
h2h_history = {}

home_forms = []
away_forms = []
h2h_forms = []

for idx, row in df.iterrows():
    home = row['HomeTeam']
    away = row['AwayTeam']
    home_goals = row['HomeGoal']
    away_goals = row['AwayGoal']
    
    if home_goals > away_goals:
        home_res = 'W'
        away_res = 'L'
        winner = home
    elif home_goals < away_goals:
        home_res = 'L'
        away_res = 'W'
        winner = away
    else:
        home_res = 'D'
        away_res = 'D'
        winner = 'D'
    
    home_last5 = team_history.get(home, [])[-5:]
    home_form = ''.join(reversed(home_last5))
    home_forms.append(home_form)
    
    away_last5 = team_history.get(away, [])[-5:]
    away_form = ''.join(reversed(away_last5))
    away_forms.append(away_form)
    
    pair = tuple(sorted([home, away]))
    h2h_last5 = h2h_history.get(pair, [])[-5:]
    
    h2h_codes = []
    for w in h2h_last5:
        if w == 'D':
            h2h_codes.append('D')
        elif w == home:
            h2h_codes.append('H')
        elif w == away:
            h2h_codes.append('A')
        else:
            h2h_codes.append('?')

    h2h_form = ''.join(reversed(h2h_codes))
    h2h_forms.append(h2h_form)
    
    team_history.setdefault(home, []).append(home_res)
    team_history.setdefault(away, []).append(away_res)
    h2h_history.setdefault(pair, []).append(winner)

df['HomeForm'] = home_forms
df['AwayForm'] = away_forms
df['H2HForm'] = h2h_forms
```
